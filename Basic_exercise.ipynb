## Exercise 01: Arithmetic Product and Conditional Logic

Practice Problem: Write a Python function that accepts two integer numbers. If the product of the two numbers is less than or equal to 1000, return their product; otherwise, return their sum.

```python
number1 = int(input("Nhập vào input 1: "))
number2 = int(input("Nhập vào input 2: "))
print(f"The result is {number1 * number2 if (number1 * number2) <= 1000 else number1 + number2}")
```
