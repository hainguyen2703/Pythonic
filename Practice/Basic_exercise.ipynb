## Exercise 02: Cumulative Sum of a Range

Practice Problem: Iterate through the first 10 numbers (0–9). In each iteration, print the current number, the previous number, and their sum.

```python
numbers = range(10)
print("Printing current and previous number sum in a range(10)")
for i in numbers:
    if i == 0:
        print("Current Number 0 Previous Number 0 Sum: 0")
    else:
        print(f"Current Number {i} Previous Number {i-1 if i > 0 else 0} Sum: {2 * i - 1}")
```

## Exercise 04: String Slicing and Substring Removal

Practice Problem: Write a function to remove characters from a string starting from index 0 up to n and return a new string.

```python
def remove_chars(text: str, index: int) -> str:
    """
    Retrun a new string from index
    """
    return text[index:]

print(remove_chars("pynative", 4))
print(remove_chars("pynative", 2))
```

## Exercise 06: Calculating Factorial with a Loop

Practice Problem: Write a program that calculates the factorial of a given number (e.g., 5!) using a for loop.

```python
number = 5
#Tính giai thừa (factorial)
result = 1
for i in range(1, number + 1):
    result *= i
print(f"The factorial of {number} is {result}")
```

## Exercise 09: Vowel Frequency Counter

Practice Problem: Write a program to count the total number of vowels (a, e, i, o, u) present in a given sentence.

```python
sentence = "Learning Python is fun!"
vowels = tuple("ueoai")
count = 0
#Đếm số nguyên âm trong sentence
for char in sentence:
    if char in vowels:
        count += 1

print("Number of vowels:", count)
```

## Exercise 16: Numerical Palindrome Check

Practice Problem: Write a program to check if a given number is a palindrome (reads the same forwards and backwards).

```python
def palindrome_check(n: int) -> bool:
    return n == int(str(n)[::-1])
#Case 1: 
number = 121
print(f"Number {number} is{"" if palindrome_check(number) == True else "not"} palindrome number")
#Case 2: 
number = 125
print(f"Number {number} is{"" if palindrome_check(number) == True else " not"} palindrome number")
```

## Exercise 17: Merging Lists with Parity Filtering

Practice Problem: Create a new list from two given lists such that the new list contains odd numbers from the first list and even numbers from the second list.

```python
list1 = [10, 20, 25, 30, 35]
list2 = [40, 45, 60, 75, 90]
#Filter odd value from first list
combine_list = list(filter(lambda x: x % 2 != 0, list1)) + list(filter(lambda x: x % 2 == 0, list2))
print(combine_list)
```

## Exercise 19: Multi-Tiered Income Tax Calculation

Practice Problem: Calculate income tax for a given income based on these rules:
1. First $10,000: 0% tax
2. Next $10,000: 10% tax
3. Remaining income: 20% tax

```python
def tax_calculation(income: int) -> float:
    tax = 0
    #First calculate tax exclusion
    income -= 10000

    #Check if the first 10% tax can be applied
    if(income > 0):
        #Check if the second 20% can be applied
        if(income - 10000 > 0):
            return 1000 + (income - 10000) * 0.2
        else:
            #Else: only 10% tax is applied
            return income * 0.1

    return tax
    
income = 13000

print(f"Total income tax to pay is {tax_calculation(income)}")
```
